```python
# ============================================
# Step 1: Import Libraries
# ============================================

import pandas as pd
import sqlite3
from google.colab import files # Import files for direct upload

# ============================================
# Step 2: Upload CSV File
# ============================================

# Direct file upload as requested by the user.
# Please select your 'FastFoodNutritionMenuV2.csv' file when prompted.
uploaded = files.upload()

# Get the name of the uploaded file
file_name = list(uploaded.keys())[0]

# Read CSV
df = pd.read_csv(file_name)

print("CSV File Uploaded Successfully!")

# ============================================
# Step 3: Data Analysis
# ============================================

print("\nFirst 5 Records")
print(df.head())

print("\nDataset Information")
df.info()

print("\nStatistical Summary")
print(df.describe())

# ============================================
# Step 4: Store Data in SQL Database
# ============================================

conn = sqlite3.connect("FastFoodNutrition.db")

df.to_sql("FastFoodMenu", conn, if_exists="replace", index=False)

print("\nData Stored Successfully in SQL Database!")

# ============================================
# Step 5: SQL Queries
# ============================================

# Display first 5 rows
query1 = "SELECT * FROM FastFoodMenu LIMIT 5;"
print("\nFirst 5 Rows from SQL")
print(pd.read_sql(query1, conn))

# Total Records
query2 = "SELECT COUNT(*) AS Total_Records FROM FastFoodMenu;"
print("\nTotal Records")
print(pd.read_sql(query2, conn))

# Display column names
print("\nColumn Names")
print(df.columns)

# Close connection
conn.close()

print("\nDatabase Connection Closed.")
```

## Importing Data from the Web

We can import data directly from a web page if it contains tabular data, often in HTML tables. For this example, I'll fetch a table from a Wikipedia page about programming languages. You can replace the URL with any page containing data in a table format.

```python
# Import data from a web page (e.g., public CSV URL)

# Example URL for a public CSV file (e.g., from GitHub or a data portal)
# This URL points to a CSV dataset of Iris flower measurements.
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'

# Read the CSV directly from the URL into a DataFrame
web_df = pd.read_csv(url)

print("\nFirst 5 Records from Web Data (Iris Dataset)")
display(web_df.head())
```

## Exporting Data to CSV and JSON Files

Now that we have data in a pandas DataFrame (`web_df`), we can easily export it to various file formats like CSV and JSON.

```python
# Export DataFrame to a CSV file
csv_filename = 'programming_languages.csv'
web_df.to_csv(csv_filename, index=False) # index=False prevents writing the DataFrame index as a column
print(f"Data successfully exported to {csv_filename}")

# Export DataFrame to a JSON file
json_filename = 'programming_languages.json'
web_df.to_json(json_filename, orient='records', indent=4) # orient='records' for list of objects, indent for pretty printing
print(f"Data successfully exported to {json_filename}")

# You can verify the creation of these files in the file browser on the left panel.
```
